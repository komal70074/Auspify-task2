```python
# Step 2: Calculate the total number of Movies and TV

count = dm["type"].value_counts()

movies = count.get("Movie", 0)
tv_shows = count.get("Tv Show", 0)
total = movies + tv_shows

print("Total Movies:", movies)
print("Total TV Shows:", tv_shows)
print("Total Content:", total)
```

```python
# Step 3: Create visualizations for content distribution.
import matplotlib.pyplot as plt 
```

```python
count.plot(kind="bar")

plt.title("Movies and TV Shows")
plt.xlabel("Content Type")
plt.ylabel("Total Count")
plt.xticks(rotation=0)

for container in plt.gca().containers:
    plt.bar_label(container, fmt="%.0f")

plt.show()
```
